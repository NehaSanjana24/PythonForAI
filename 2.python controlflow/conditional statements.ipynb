```python
## Conditional Statements 
## Conditional statements are used to perform different actions based on different conditions.
## if, elif, and else are the keywords used for conditional statements in Python.q
```

```python
##if statement is used to test a specific condition. If the condition is true, the block of code inside the if statement is executed. If the condition is false, the code inside the if statement is skipped.
age = 16
if age >= 18:
    print("You are eligible to vote.")
print("Thank you for checking your eligibility.")
```

```python
## else statement is used to execute a block of code when the condition in the if statement is false.
age = 22
if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
print("Thank you for checking your eligibility.")
```

```python
## elif statement is used to test multiple conditions. It allows you to check for additional conditions if the previous conditions are false.
age = 19
if age >= 18 :
    print("You are eligible to vote and drive.")
elif age >= 16:
    print("You are eligible to drive.")
else:
    print("You are not eligible to vote or drive.")
```

```python
##Nested conditional statements are used when you have multiple conditions to check. You can nest if statements inside other if statements to create more complex logic.
##number even,odd,negative,positive

num=int(input("Enter a number: "))
if num>0:
    if num%2==0:
        print(f"the number {num} is even and positive")
    else:
        print(f"the number {num} is odd and positive")
elif num<0:
    if num%2==0:
        print(f"the number {num} is negative and even")
    else:
        print(f"the number {num} is negative and odd")
elif num==0:
   print(f"the number {num} is zero")
else:
    print("enter the valid number")
```

```python
## practical example
## Determine if a yesr is leapyeaer or not
year=int(input("Enter the year:"))
if (year%400==0 or (year%100!=0 and year%4==0)):
    print(f"{year} is leap year.")
else:
    print(f"{year} is not a leap year.")
```

```python
## Simple Calculator
# Simple Calculator
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
operation = input("Select operation (+, -, *, /): ").strip()

if operation == '+':
    result = a + b
elif operation == '-':
    result = a - b
elif operation == '*':
    result = a * b
elif operation == '/':
    if b != 0:
        result = a / b
    else:
        result = "Error (Division by zero)"
else:
    result = "Invalid operation"

print(f"{a} {operation} {b} = {result}")
```
