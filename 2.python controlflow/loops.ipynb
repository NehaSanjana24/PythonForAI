```python
##for loop
## The for loop is used to iterate over a sequence (such as a list, tuple, string, or range) and execute a block of code for each item in the sequence. It allows you to perform repetitive tasks efficiently.

range(5)  # This creates a sequence of numbers from 0 to 4
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,100,2):
    print(i)
```

```python
for i in range(100,0,-5):
    print(i)
```

```python
##String Iteration
str="Neha Sanjana"
for i in str:
    print(i)
    
```

```python
## While loop
## The while loop is used to repeatedly execute a block of code as long as a specified condition is true. It allows you to create loops that continue until a certain condition is no longer met.

count=1
while count<=5:
    print(count)
    count+=1

```

```python
count=0
while count%2==0:
    print(count)
    count+=1
```

```python
## Loop Control Statements
## The break statement is used to exit a loop prematurely, even if the loop condition is still true.
## The continue statement is used to skip the current iteration of a loop and move to the next iteration.   
## The pass statement is a placeholder that does nothing and is often used as a temporary placeholder for code that will be implemented later.
```

```python
##Break Statement
for i in range(1,10):
    if i==7:
        break
    print(i)
```

```python
##Continue Statement
for i in range (1,10):
    if i==4:
        continue
    print(i)
```

```python
for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
##Pass Statement
for i in range (1,100):
    if i%3==0:
        print(f"The number {i} is divisible by 3")
        pass
    else:
        print(i)
```

```python
##Nested Loops
## Nested loops are loops that are placed inside another loop. The inner loop is executed for each iteration of the outer loop. Nested loops are useful when you need to perform operations on multi-dimensional data structures or when you want to iterate over combinations of elements.
for i in range(1,4):
    for j in range (1,4):
        print(f"i={i}, j={j}")
```

```python
##Examples- calculate the sum of first n natural numbers using for loop and while loop

##For loop

n=int(input())
sum=0
for i in range(1,n+1):
    sum+=i
print(f"Sum of first {n} natural numbers is {sum}")
```

```python
##While loop
n=int(input())
sum=0
count=1
while n>=count:
    sum+=count
    count+=1
print(f"Sum of first {n} natural numbers is {sum}")
```

```python
##Example -- prime number between given range

##For loop
start=int(input("Enter the start of the range: "))
end=int(input("Enter the end of the range: "))
for i in range(start,end+1):
    if i >1:
        for j in range(2,i):
            if i%j==0:
                break
        else:
            print(i)
```
