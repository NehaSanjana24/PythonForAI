```python
## Naming conventions
## Variable names can contain letters, numbers, and underscores, but they cannot start with a number. Variable names are case-sensitive, meaning that "myVariable" and "myvariable" would be considered two different variables.
## They should not be a reserved keyword in Python, such as "if", "else", "while", etc. It is recommended to use descriptive names for variables to make the code more readable and maintainable.
## They must start with a letter or an underscore, and can be followed by letters, numbers, or underscores. They cannot contain spaces or special characters.

#valid variable names
first_name = "John"
last_name = "Doe"
```

```python
##Understanding Variable types
##Python is dynamically typed, which means that you don't need to declare the type of a variable when you create it. The type of a variable is determined by the value it is assigned. You can use the `type()` function to check the type of a variable.
age=19
height=4.9
name="neha"
is_student=True
print("Type of age:",type(age))
print("Type of height:",type(height))
print("Type of name:",type(name))
print("Type of is_student:",type(is_student))
```

```python
## Type checking and type conversion
## You can check the type of a variable using the `type()` function. You can also convert a variable from one type to another using type conversion functions like `int()`, `float()
type(height)
```

```python
age=25
print(type(age))  
#Type conversion
age_str=str(age)
print(age_str)
print(type(age_str))
```

```python
age='25'
print(type(age))
print(type(int(age)))  #converting string to integer
```

```python
name="neha"
print(type(name))
print(type(int(name)))  #converting string to integer not possible as it will give an error because the string "neha" cannot be converted to an integer.
```

```python
height=4.9
print(type(height))
print(height)
print(type(int(height)))  #converting float to integer
print(int(height))
```

```python
##Dynamic Typing
'''Python is dynamically typed, which means that you can change the type of a variable by assigning a new value to it. For example, you can assign an integer value to a variable and then
    assign a string value to the same variable later in the code. This is different from statically typed languages, where the type of a variable is fixed at the time of declaration and cannot be changed later.'''

var=10 #int
print(var,type(var))
var="Hello" #string
print(var,type(var))
var=3.14 #float
print(var,type(var))
```

```python
##input() function
age=input("Enter your age: ")   
print(age)  #input() function always returns a string, so the type of the variable will be str. If you want to convert the input to a different type, you can use type conversion functions like int() or float().
print(type(age))
```
