```python
## Data Types in Python
## Python has several built-in data types that are used to store different kinds of data. 
## data types are a classification of data items. It represents the kind of value that tells what operations can be performed on a particular data.
## Impoertance of Data Types
## Data types are important because they determine how the data is stored in memory, how much space it occupies, and what operations can be performed on it. They also help in defining the behavior of variables and functions in a program. Using the correct data type can lead to more efficient code and can prevent errors that may arise from type mismatches.
```

```python
## Integers: Integers are whole numbers without a decimal point. They can be positive, negative, or zero. In Python, integers are represented by the int data type.
age = 25
type(age)  
```

```python
##float: Floats are numbers that have a decimal point. They can also be positive, negative, or zero. In Python, floats are represented by the float data type.
temperature = 36.6
type(temperature)  
```

```python
## Strings: Strings are sequences of characters enclosed in single quotes, double quotes, or triple quotes. In Python, strings are represented by the str data type.
name = "John Doe"
type(name)
```

```python
## Boolean: Booleans are a data type that can have one of two values: True or False. In Python, booleans are represented by the bool data type.
is_student = True
type(is_student)
```

```python
is_student=bool()
type(is_student)
```

```python
a=10
b=10
print(a==b)
type(a==b)
```
